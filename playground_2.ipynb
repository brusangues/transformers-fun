```python
from pathlib import Path
folder_raw = "data/tolkien_raw_pt1"
folder_txt = "data/tolkien_txt_pt1"

Path(folder_txt).mkdir(parents=True, exist_ok=True)
print(f"{folder_raw=}")
epubs = list(Path(folder_raw).glob("*.epub"))
print(f"{len(epubs)=}")
epubs[:2]
```

```python
import ebooklib
from ebooklib import epub
from bs4 import BeautifulSoup

def epub_to_txt(path_epub, folder_txt):
    book = epub.read_epub(path_epub)
    title = book.get_metadata('DC', 'title')[0][0]

    content = []
    for item in book.get_items():
        if item.get_type() == ebooklib.ITEM_DOCUMENT:
            soup = BeautifulSoup(item.get_content(), 'html.parser')
            text = soup.get_text()
            content.append(text)

    path_txt = folder_txt + "/" + title + ".txt"
    # if not Path(path_txt).exists():
    #     with open(path_txt, 'w', encoding='utf-8') as f:
    #         f.write('\n'.join(content))
    # else:
    #     print(f"Skipping {path_txt}, already exists.")

    metadata = {
        'title': title,
        'author': book.get_metadata('DC', 'creator')[0][0],
        "path_raw": str(path_epub),
        "path_txt": path_txt,
        "text_raw_len": len('\n'.join(content)),
    }
    print(f"{metadata=}")

    return metadata


metadatas = []
for file in epubs:
    print(f"\nConverting {file}...")
    try:
        metadata = epub_to_txt(file, folder_txt)
        metadatas.append(metadata)
    except Exception as e:
        print(f"Error converting {file}: {e}")
```

```python
import pandas as pd

df = pd.DataFrame(metadatas)
```

```python
df_full = pd.read_parquet("data/df_full_v21.pq")
```

```python
text_full = "\n\n\n".join(df_full.query("split=='train'").sample(frac=1).text_clean.to_list())
```

```python
print(f"{text_full[:2_000]=}")
print(f"{len(text_full)=}")
```

```python
batch_size = int(7e7)

def batch_iterator():
    for i in range(0, len(text_full), batch_size):
        yield text_full[i:i+batch_size]

len_batches = len(list(batch_iterator()))
lens_batches = [len(b) for b in batch_iterator()]
print(f"{len_batches=}")
print(f"{lens_batches=}")
for i, b in enumerate(batch_iterator()):
    print(i, b[:50])
```

```python
from transformers import AutoTokenizer
 
tokenizer = AutoTokenizer.from_pretrained("gpt2")
tokenizer.is_fast
```

```python
new_tokenizer = tokenizer.train_new_from_iterator(batch_iterator(), vocab_size=50257)
```

```python
new_tokenizer.save_pretrained("artifacts/tokenizers/gpt2_ptbr_50k_v2")
```

```python
from tqdm import tqdm

tqdm.pandas()
print("Encoding input texts...")
encode = lambda s: new_tokenizer(s, truncation=False).input_ids
df_full["text_encoded"] = df_full.text_clean.progress_apply(lambda x: encode(x))
df_full["text_encoded_len"] =  df_full.text_encoded.apply(len)
print(f"{df_full.text_encoded_len.describe()=}")
```

```python
print("\nChars/Tokens statistics:")
len_corpus = int(df_full["text_clean_len"].sum()) / 1e6
len_splits = (df_full.groupby("split").text_len.sum() / 1e6).to_dict()
print("Chars:")
print(f"{len_corpus=} {len_splits=} M chars")

len_corpus = int(df_full["text_encoded_len"].sum()) / 1e6
len_splits = (df_full.groupby("split").text_encoded_len.sum() / 1e6).to_dict()
print("Tokens:")
print(f"{len_corpus=} {len_splits=} M tokens")

compression_ratio = (df_full["text_len"].sum()) / (df_full["text_encoded_len"].sum())
print(f"Overall compression ratio (chars/tokens): {compression_ratio:.2f}")
```
