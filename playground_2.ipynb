```python
from pathlib import Path
folder_raw = "data/tolkien_raw_pt1"
folder_txt = "data/tolkien_txt_pt1"

Path(folder_txt).mkdir(parents=True, exist_ok=True)
print(f"{folder_raw=}")
epubs = list(Path(folder_raw).glob("*.epub"))
print(f"{len(epubs)=}")
epubs[:2]
```

```python
import ebooklib
from ebooklib import epub
from bs4 import BeautifulSoup

def epub_to_txt(path_epub, folder_txt):
    book = epub.read_epub(path_epub)
    title = book.get_metadata('DC', 'title')[0][0]

    content = []
    for item in book.get_items():
        if item.get_type() == ebooklib.ITEM_DOCUMENT:
            soup = BeautifulSoup(item.get_content(), 'html.parser')
            text = soup.get_text()
            content.append(text)

    path_txt = folder_txt + "/" + title + ".txt"
    if not Path(path_txt).exists():
        with open(path_txt, 'w', encoding='utf-8') as f:
            f.write('\n'.join(content))
    else:
        print(f"Skipping {path_txt}, already exists.")

    metadata = {
        'title': title,
        'author': book.get_metadata('DC', 'creator')[0][0],
        "path_epub": str(path_epub),
        "path_txt": path_txt,
    }
    print(f"{metadata=}")

    return metadata


metadatas = []
for file in epubs:
    print(f"\nConverting {file}...")
    try:
        metadata = epub_to_txt(file, folder_txt)
        metadatas.append(metadata)
    except Exception as e:
        print(f"Error converting {file}: {e}")
```

```python
import pandas as pd

df = pd.DataFrame(metadatas)
```
